```python
# Load Ontology definision file 
#https://cedric.cnam.fr/~hamdif/ontologies/files/PersonLink.html
from rdflib import Graph, Namespace, RDF, RDFS, OWL
g = Graph()
g.parse("https://cedric.cnam.fr/~hamdif/ontologies/files/PersonLink.owl",format='xml')

def extract_object_properties(graph):
    object_properties = []
    for s, p, o in graph.triples((None, RDF.type, OWL.ObjectProperty)):
        for label in graph.objects(s, RDFS.label):
            if label.language == 'en':
                object_properties.append((s, label))
                break  # If we found an English label, no need to check further
    return object_properties

# Extract and print object properties
object_properties = extract_object_properties(g)
print("Object Properties in the Ontology:")
for prop, label in object_properties:
    print(f"{label} ({prop})")
```

## Interpretation of a rule 
FatherOf(?x, ?y), FatherOf(?y, ?z) -> GrandFatherOf(?x, ?a)
### Condition
 - FatherOf(?x, ?y): This indicates that ?x is the father of ?y.
 - FatherOf(?y, ?z): This indicates that ?y is the father of ?z.

### Conclusion
 - GrandFatherOf(?x, ?a): This is the conclusion we want to draw, stating that ?x is the grandfather of ?a.

```python
from owlready2 import *
import re

# Load the ontology
ontology_path = "data\corrected_ontology.owl"
onto = get_ontology(ontology_path).load(reload=True)

# Regular expression to extract variables
variable_pattern = re.compile(r'\?(\w+)')

# Function to extract variables from a rule atom
def extract_variables(atom):
    return set(variable_pattern.findall(str(atom)))

# Function to get the label of an entity in English
def get_label(entity):
    if entity.label:
        for label in entity.label:
            if label.lang == "en":
                return label
        return entity.label.first()
    else:
        return entity.name

# Function to convert a rule atom to a labeled string
def convert_atom_to_label(atom):
    parts = atom.split("(")
    predicate = parts[0]
    variables = parts[1][:-1]  # remove the trailing ')'
    predicate_label = get_label(onto[predicate])
    return f"{predicate_label}({variables})"

# Function to convert a rule to a labeled string
def convert_rule_to_labels(rule):
    body = ", ".join([convert_atom_to_label(str(atom)) for atom in rule.body])
    head = ", ".join([convert_atom_to_label(str(atom)) for atom in rule.head])
    return f"{body} -> {head}"

# Function to print all rules in the ontology with English labels
def print_all_rules_with_labels(onto):
    for rule in onto.rules():
        labeled_rule = convert_rule_to_labels(rule)
        print(f"Rule: {labeled_rule}")

# Print all rules with English names
print_all_rules_with_labels(onto)
```

```python
# Function to get the English label of an entity
def get_label(entity):
    
    if hasattr(entity, 'label') and entity.label:
        for i in entity.label:
            if i.lang == 'en':
                return i
        
    else:
        return entity.name

# Function to print all relations in the ontology with English names
def print_all_relations(ontology):
    print("Listing all relations in the ontology:\n")
    
    # Print object properties and their relations
    print("Object Properties:")
    for obj_property in ontology.object_properties():
        property_label = get_label(obj_property)
        print(f"Property: {property_label}")
        for value in obj_property.get_relations():
            subject_label = get_label(value[0])
            object_label = get_label(value[1])
            print(f"  {subject_label} -- {property_label} --> {object_label}")



# Print all relations
print_all_relations(onto)
```
