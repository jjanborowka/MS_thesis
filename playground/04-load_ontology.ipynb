```python
from rdflib import Graph, RDF, RDFS

# Load the ontology into an RDF graph
g = Graph()
ttl_url = "https://raw.githubusercontent.com/blokhin/genealogical-trees/master/data/header.ttl"
g.parse(ttl_url, format="turtle")

g
```

```python
from rdflib import Graph, Namespace, BNode, URIRef
from rdflib.namespace import RDF, RDFS, OWL
namespaces = {
    "rdf": RDF,
    "rdfs": RDFS,
    "owl": OWL,
    "ex": Namespace("http://www.example.com/genealogy.owl#")
}

def make_readable(term):
    """Convert rdflib terms into a readable format."""
    if isinstance(term, BNode):
        return "BlankNode"
    elif isinstance(term, URIRef):
        for prefix, ns in namespaces.items():
            if term.startswith(ns):
                return f"{prefix}:{term.replace(ns, '')}"
        return term.split('/')[-1]  # Fallback to last part of the URI
    else:
        return str(term)  # Handle Literals or other types

# Print the first triple in a human-readable format
for subj, pred, obj in g:
    print("First triple (human-readable):")
    print("Subject:", make_readable(subj))
    # print("Predicate:", make_readable(pred))
    print(pred)
    print("Object:", make_readable(obj))
    break  # Stop after the first triple
```
